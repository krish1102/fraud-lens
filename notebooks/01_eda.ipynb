# FraudLens — 01: Exploratory Data Analysis (EDA)
## IEEE-CIS Benchmark Dataset & Online Transaction Ingestion
This notebook demonstrates exploratory analysis, missing value distributions, class imbalance, and transaction amounts.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Load development benchmark data
data_path = Path('../data/raw/development_transactions.csv')
if not data_path.exists():
    data_path = Path('../data/demo/sample_ecommerce_transactions.csv')
df = pd.read_csv(data_path)
print(f"Dataset shape: {df.shape}")
df.head()
```

```python
# Inspect class distribution and imbalance
if 'isFraud' in df.columns:
    print("Class distribution:")
    print(df['isFraud'].value_counts(normalize=True))
    
# Summary statistics of amount
df['amount'].describe()
```

```python
# Visualizing amount distribution
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.histplot(df['amount'], bins=40, kde=True, color='skyblue')
plt.title('Transaction Amount Distribution')

plt.subplot(1, 2, 2)
sns.histplot(np.log1p(df['amount']), bins=40, kde=True, color='salmon')
plt.title('Log(1 + Amount) Distribution')
plt.tight_layout()
plt.show()
```
